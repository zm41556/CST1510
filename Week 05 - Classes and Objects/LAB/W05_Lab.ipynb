### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6: Classify records by percentage of their limits.
class Record:
    """Represent one labeled value and the limit used to assess it."""
    def __init__(self, label, value, limit):
        """Initialize the object with the supplied values."""
        self.label = label
        self.value = value
        self.limit = limit

    def status(self):
        """Return this record's status relative to its limit."""
        percent = (self.value/self.limit) * 100
        if percent >= 100:
            result = "OVER LIMIT"
        elif percent >= 90:
            result = "WARNING"
        else:
            result = "OK"
        return result

r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)

print(f" Record 1 Status: {r1.status()}")
print(f" Record 2 Status: {r2.status()}")
print(f" Record 3 Status: {r3.status()}")
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.

**Expected output:** `87.0`, then `OK`

````python
# D7: Reuse percent() to determine the record status.
class Record:
    """Represent one labeled value and the limit used to assess it."""
    def __init__(self, label, value, limit):
        """Initialize the object with the supplied values."""
        self.label = label
        self.value = value
        self.limit = limit

    def status(self):
        """Return this record's status relative to its limit."""
        percent = self.percent()
        if percent >= 100:
            result = "OVER LIMIT"
        elif percent >= 90:
            result = "WARNING"
        else:
            result = "OK"
        return result
    
    def percent(self):
        """Return the value as a percentage of the limit."""
        percent = (self.value / self.limit) * 100
        return percent

record = Record("R-001", 87.0, 100.0)

print(f"Record Percent: {record.percent()}")
print(f"Record Status: {record.status()}")
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8: Track over-limit checks on one Checker instance.
class Checker:
    """Track values that exceed their limits."""
    def __init__(self):
        """Initialize the object with the supplied values."""
        self.over_limit_count = 0

    def check(self, value, limit):
        """Increase the over-limit count when value exceeds limit."""
        if value > limit:
            self.over_limit_count += 1

checker = Checker()
checker.check(87, 100)
checker.check(120, 100)
checker.check(130, 100)
print(checker.over_limit_count)
````

### D9. A list of objects

1. Using your `Record` class from D7, create a list of three records:
   `R-001` (87.0), `R-002` (120.0) and `R-003` (45.0), each with a limit of 100.0.
2. Loop over the list and print each record's label and status.

**Expected output:**

````python
# D9: Build and process a list of Record objects.
class Record:
    """Represent one labeled value and the limit used to assess it."""
    def __init__(self, label, value, limit):
        """Initialize the object with the supplied values."""
        self.label = label
        self.value = value
        self.limit = limit

    def status(self):
        """Return this record's status relative to its limit."""
        percent = self.percent()
        if percent >= 100:
            result = "OVER LIMIT"
        elif percent >= 90:
            result = "WARNING"
        else:
            result = "OK"
        return result
    
    def percent(self):
        """Return the value as a percentage of the limit."""
        percent = (self.value / self.limit) * 100
        return percent

records = [
    Record("R-001", 87.0, 100.0),
    Record("R-002", 120.0, 100.0),
    Record("R-003", 45.0, 100.0)
]


for record in records:
    print(f"{record.label} {record.status()}")
````

### D10. A class-level counter

1. Define a new, small class `Counted`. Directly under the `class` line (not inside `__init__`),
   add `count = 0`.
2. In `__init__(self)`, add 1 to `Counted.count`.
3. Create three `Counted` objects, then print `Counted.count`.

**Expected output:** `3`

Add a comment explaining why the count is shared by every object instead of each object having its own.

````python
# D10: Share an instance count through a class attribute.
class Counted:
    """Count how many instances of this class have been created."""
    count = 0
    def __init__(self):
        """Initialize the object with the supplied values."""
        Counted.count += 1

c1 = Counted()
c2 = Counted()
c3 = Counted()

print(Counted.count)
# Because count belongs to the class, not instance, so it is shared across all objects.
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11: Load rows from the sample file as Record objects.

class Record:
    """Represent one labeled value and the limit used to assess it."""
    def __init__(self, label, value, limit):
        """Initialize the object with the supplied values."""
        self.label = label
        self.value = value
        self.limit = limit

    def status(self):
        """Return this record's status relative to its limit."""
        percent = self.percent()
        if percent >= 100:
            result = "OVER LIMIT"
        elif percent >= 90:
            result = "WARNING"
        else:
            result = "OK"
        return result
    
    def percent(self):
        """Return the value as a percentage of the limit."""
        percent = (self.value / self.limit) * 100
        return percent

    
def load_records(path):
    """Read label,value,limit lines from a file and return a list of Records."""
    records_list = []
    with open(path, "r") as f:
        for line in f:
            if not line.strip():
                continue
            label, value, limit = line.strip().split(",")
            item  = Record(label, float(value), float(limit))
            records_list.append(item)
    return records_list

from pathlib import Path

# Resolve the sample file from the notebook, Week 05, or repository root.
record_path = next(
    path for path in (
        Path("records_sample.txt"),
        Path("LAB/records_sample.txt"),
        Path("Week 05 - Classes and Objects/LAB/records_sample.txt"),
    )
    if path.is_file()
)
records = load_records(record_path)

for record in records:

    print(f"{record.label} {record.status()}")
````

Add a method `is_over(self)` to `Record` that returns `True` or `False`. Load `records_sample.txt` with
`load_records` from D11, use `is_over()` to count how many records are over their limit, and print a
one-line summary.

**Expected output:** `1 of 3 records are OVER LIMIT`

````python
# C1: Count records whose values exceed their limits.
class Record:
    """Represent one labeled value and the limit used to assess it."""
    over_limit_count = 0
    def __init__(self, label, value, limit):
        """Initialize the object with the supplied values."""
        self.label = label
        self.value = value
        self.limit = limit
    
    def status(self):
        """Return this record's status relative to its limit."""
        percent = (self.value / self.limit) * 100
        if percent >= 100:
            return "OVER LIMIT"
        if percent >= 90:
            return "WARNING"
        return "OK"

    def is_over(self):
        """Return whether this record exceeds its limit."""
        return self.status() == "OVER LIMIT"

def load_records(path):
    """Read label,value,limit lines from a file and return a list of Records."""
    records_list = []
    with open(path, "r") as f:
        for line in f:
            if not line.strip():
                continue
            label, value, limit = line.strip().split(",")
            item  = Record(label, float(value), float(limit))
            records_list.append(item)
    return records_list

from pathlib import Path

# Resolve the sample file from the notebook, Week 05, or repository root.
record_path = next(
    path for path in (
        Path("records_sample.txt"),
        Path("LAB/records_sample.txt"),
        Path("Week 05 - Classes and Objects/LAB/records_sample.txt"),
    )
    if path.is_file()
)
records = load_records(record_path)


over_count = 0
for record in records:
    if record.is_over():
        over_count += 1

print(f"{over_count} of {len(records)} are OVER LIMIT")
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2: Lock an account after repeated failed logins.
class Account:
    """Track failed login attempts for one account."""
    def __init__(self, username):
        """Initialize the object with the supplied values."""
        self.username = username
        self.failed_logins = 0

    def fail(self):
        """Record one failed login attempt."""
        self.failed_logins += 1

    def locked(self):
        """Return whether the account has at least three failed logins."""
        return self.failed_logins >= 3

anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())

ben = Account("ben")
ben.fail()
print(ben.locked())
````
