## 1. Writing, with `with`

`with ... as f:` opens the file, hands you `f` to work with, and closes it
automatically when the indented block ends — even if something goes wrong partway
through.

````python
# The "w" mode creates the file or replaces its contents.
with open("output.txt", "w") as f:
    f.write("R-001\n")
    f.write("R-002\n")
````

```
with open("output.txt", "w") as f:
       │
       ▼
  file is open, f is your handle
       │
       ▼
  ... f.write(...) ...
       │
       ▼
  block ends  ──▶  file closes automatically, even if an error happened
```

## 2. Reading

No mode given defaults to `"r"` (read). `"w"` overwrites the whole file every time
you open it that way — `"a"` appends instead.

````python
with open("output.txt") as f:
    text = f.read()

print(text)
````

## 3. Reading line by line

Looping over an open file gives you one line at a time — the natural way to
process many records. Each line keeps its trailing `\n` — `.strip()` removes it.

````python
with open("output.txt") as f:
    for line in f:
        print(line.strip())
````

## 4. Splitting a line into fields

`.split(",")` turns one line of text into a list of values — still text, so
convert what needs converting.

````python
line = "R-001,87,100"
label, value, limit = line.split(",")
value = float(value)
limit = float(limit)
print(label, value, limit)
````

## Topic 3 Challenge

Write a function `load_records(path)` that reads a file of `label,value,limit`
lines and returns a list of dicts — one dict per line, shaped like
`{"label": ..., "value": ..., "limit": ...}`.

````python
# your code here

 #this function read the file and return a list 'record' of dict 
def load_records(path):
    """Read label,value,limit lines from a file and return a list of dicts."""
    records = [] #the list that the dict will be added to
    with open(path, "r") as file: #reading the file 
        for line in file:
            label, value, limit = line.strip().split(",") #spliting data from the file
            records.append({ # turning data from the file into a dict for every line and append it to the records list
                "label": label,
                "value": float(value),
                "limit": float(limit)
            })
             
    
    return records #return the record list 

records = load_records("record.txt") # define records as 'load_records' function because this is what the function return
for record in records: 
    print(record)   #display the file content in a list of dict
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| opening a file that does not exist | `FileNotFoundError` | check the path — is it relative to where you are running from? (or catch it — Topic 4) |
| forgetting `.strip()` before comparing or splitting a line | odd, hard-to-spot mismatches | every line from a file carries a trailing `\n` |
| using `"w"` when you meant `"a"` | your previous content is gone | `"w"` overwrites, `"a"` appends — pick deliberately |
| forgetting to convert after `.split()` | `TypeError` when you try to calculate | everything from `.split()` is still text |
